```python
# Step 1: Identify unique parties present in df2 for each 'pc' number
valid_parties_per_pc = ms.groupby('State')['Caste'].unique().to_dict()

# Step 2: Map parties in df1 that are not present in df2 for each 'pc' number
def map_party(row):
    if row['Caste Name'] in valid_parties_per_pc.get(row['State'], []):
        return row['Caste Name']
    else:
        return 'Others'

# Step 3: Filter retro DataFrame to include only those rows where 'pc_id' is present in ms
filtered_retro = caste[caste['State'].isin(valid_parties_per_pc.keys())]

# Step 4: Apply mapping function to each row in filtered retro DataFrame and create new column
filtered_retro['Caste Name'] = filtered_retro.apply(map_party, axis=1)
others_vote_share = filtered_retro[filtered_retro['Caste Name'] == 'Others'].groupby('State')['%'].sum()

# Create a new DataFrame for 'Others' with aggregated vote share
others_df = pd.DataFrame({
    'State': others_vote_share.index,
    'Caste Name': 'Others',
    '%': others_vote_share.values
})

# Append the 'Others' DataFrame to the original DataFrame
caste = pd.concat([filtered_retro[filtered_retro['Caste Name'] != 'Others'], others_df], ignore_index=True)

caste.shape
```
